# Caso Práctico: Neurona de McCulloch y Pitts

## Implementación de la MPNeuron

```python
# Importa la librería NumPy y le asigna el alias 'np' para operaciones numéricas y manejo de arreglos
import numpy as np

# Define la clase que representa a la neurona de McCulloch-Pitts
class MPNeuron:  
    def __init__(self):  # Método constructor que inicializa los atributos de la clase al crear una instancia
        self.threshold = None  # Inicializa el atributo 'threshold' (umbral) como vacío o nulo por defecto

    # Define un método para evaluar una sola muestra de entrada (un vector de características)
    def model(self, x):  
        # input: [1, 0, 1, 0] [x1, x2, .., xn]
        z = sum(x)  # Suma todos los valores de las características de la entrada 'x'
        return (z >= self.threshold)  # Retorna True si la suma es mayor o igual al umbral, de lo contrario False

    # Define un método para hacer predicciones en lote sobre múltiples muestras (una matriz)
    def predict(self, X):  
        # input: [[1, 0, 1, 0], [1, 0, 1, 1]]  
        Y = []  # Crea una lista vacía para almacenar los resultados de las predicciones
        for x in X:  # Recorre iterativamente cada muestra individual 'x' dentro del conjunto de datos 'X'
            result = self.model(x)  # Evalúa la muestra actual llamando al método 'model'
            Y.append(result)  # Añade el resultado booleano obtenido a la lista 'Y'
        return np.array(Y)  # Convierte la lista de resultados en un arreglo de NumPy y lo retorna
```

```python
# Instanciamos la neurona
mp_neuron = MPNeuron()
```

```python
# Establecemos un threshold
mp_neuron.threshold = 4
```

```python
# Evaluamos diferentes casos de uso
mp_neuron.predict([[1, 0, 0, 0],[1, 1, 1, 1], []])
```
