### SELECCIÓN DE ATRIBUTOS 

```python
import numpy as np
from sklearn.feature_selection import VarianceThreshold
```

```python
# Carga de datos.
X = [[0, 0, 1], [0, 1, 0], [1, 0, 0],
     [0, 1, 1], [0, 1, 0], [0, 1, 1]]
print("Matriz 6 x 3")
print(X)
```

```python
# Limpieza de datos: selección de atributos.
# fijamos un umbral (th) para decidir qué porcentaje mínimo deben variar los valores de una columna para considerarse
# una columna relevante a la hora de mantenerla en nuestro estudio
th = 0.2 #valor que se considera de acuerdo a criterio
sel = VarianceThreshold(th)
sel.fit(X)

# Varianzas calculadas para cada columna
varianzas = sel.variances_

# Decisión (True = se selecciona, False = se rechaza)
seleccionado = sel.get_support()

# Resultados de cada atributo/columna
for i, (var, sel_status) in enumerate(zip(varianzas, seleccionado)):
    estado = "Seleccionado (Se queda)" if sel_status else "Eliminado (Baja varianza)"
    print(f"Columna {i}: Varianza = {var:.4f} -> {estado}")
    
print("\n Matriz con atributos seleccionados")
print(sel.fit_transform(X))
```
